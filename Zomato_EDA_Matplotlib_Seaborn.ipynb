# EDA on Zomato Data: Matplotlib & Seaborn (30-min Session)

**Dataset:** `zomato_small.csv` (20 restaurants, City, Cuisine, Rating, Votes, Cost, Online Order, Table Booking)

**Goal:** See 7 core plot types, each in 1–2 lines of code. Talk about *what the plot answers*, not syntax.


## 0. Setup (30 sec)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
df = pd.read_csv("zomato_small.csv")
df.head()
```

## 1. Countplot "How many of each category?"
**Question:** Which cuisine is most common?

```python
sns.countplot(x="Cuisine", data = df)
```

```python
sns.countplot(x="Cuisine", data=df)
plt.xticks(rotation=45)
```

```python
sns.countplot(data = df, x = "Cuisine")
plt.xticks(rotation=45)
```

```python
sns.countplot(data=df, x="Cuisine", order=df["Cuisine"].value_counts().index)
plt.xticks(rotation=45)
plt.title("Restaurant Count by Cuisine")
plt.show()
```

## 2. Histogram / KDE "How is a number distributed?"
**Question:** How are ratings spread out?

```python
sns.histplot(data=df, x="Rating", bins =8)
```

```python
sns.histplot(data=df, x="Rating", bins=10, kde =True)
```

```python
sns.histplot(data=df, x="Rating", bins=8, kde=True)
plt.title("Distribution of Ratings")
plt.show()
```

## 3. Boxplot "Spread + outliers across groups"
**Question:** Does cost-for-two vary by city? Any outliers?

```python
sns.boxplot(data=df, x="City", y="Cost_for_two")
```

```python
sns.boxplot(data=df, x="City", y="Cost_for_two")
```

```python
sns.boxplot(data=df, x="City", y="Cost_for_two")
plt.title("Cost for Two by City")
plt.show()
```

## 4. Barplot "Average of a number, per category"
**Question:** Which city has the highest average rating?

```python
sns.barplot(data = df, x="City", y="Rating")
```

```python
sns.barplot(data=df, x="City", y="Rating")
```

```python
sns.barplot(data=df, x="City", y="Rating", estimator="mean")
plt.title("Average Rating by City")
plt.show()
```

## 5. Scatterplot "Relationship between two numbers"
**Question:** Do more votes mean higher rating? (add hue for a 3rd variable)

```python
sns.scatterplot(data=df, x="Votes", y="Rating")
plt.title("Votes vs Rating (colored by Online Order)")
plt.show()
```

## 6. Heatmap "Correlation between all numeric columns at once"
**Question:** Are cost, votes, and rating related?

```python
corr = df[["Rating", "Votes", "Cost_for_two"]].corr()
sns.heatmap(corr, annot=False, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## 7. Pairplot "All pairwise relationships in one shot"
**Question:** Quick overview before deeper analysis great EDA starting point.

```python
sns.pairplot(df, vars=["Rating", "Votes", "Cost_for_two"], hue="Online_order")
plt.show()
```

## Quick Reference "Which plot do I use?"

| Question you're asking | Plot | Function |
|---|---|---|
| How many per category? | Countplot | `sns.countplot()` |
| How is one number distributed? | Histogram/KDE | `sns.histplot()` |
| Spread & outliers across groups? | Boxplot | `sns.boxplot()` |
| Average of a number per group? | Barplot | `sns.barplot()` |
| Relationship between 2 numbers? | Scatterplot | `sns.scatterplot()` |
| Correlation of many numbers at once? | Heatmap | `sns.heatmap()` |
| Quick overview of everything? | Pairplot | `sns.pairplot()` |

**Pacing tip (30 min):** ~2 min setup, ~3–4 min per plot (run → ask "what does this tell us?" → move on), ~2 min wrap-up on the cheat table.
