```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
X = iris.data
```

```python
X
```

```python
y = iris.target
```

```python
y
```

```python
feature_names = iris.feature_names
feature_names
```

```python
X.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled.mean(axis = 0)
```

```python
X_scaled.std(axis = 0)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca.shape
```

```python
X_pca
```

```python
pca.explained_variance_ratio_
```

```python
pca.explained_variance_ratio_.sum()
```

```python
plt.figure(figsize=(9,7))

for class_id, species in enumerate(iris.target_names):
    mask = y == class_id

    plt.scatter(
        X_pca[mask,0],
        X_pca[mask,1],
        label = species
    )

plt.title("PCA: iris data reduced from 4D to 2D")
plt.xlabel("Principle component 1")
plt.ylabel("Principle component 2")
plt.legend()
```

```python
explained_variance = pca.explained_variance_ratio_

cumlative_variance = np.cumsum(explained_variance)
```

```python
cumlative_variance
```

```python
full_pca = PCA()

full_pca.fit(X_scaled)

cumlative = np.cumsum(full_pca.explained_variance_ratio_)
```

```python
cumlative
```

```python
plt.figure(figsize=(9,7))

plt.plot(
    range(1, len(cumlative) + 1),
    cumlative,
    marker = 'o'
)

plt.xlabel("Number of components")
plt.ylabel("Cumlative explained variance ration")
plt.title("Choosing principle components for PCA")
```
