**Q22.** Build a ColumnTransformer (impute+scale numerics, one-hot encode categoricals) from raw data.

```python
raw = pd.DataFrame(data).drop(columns=["Name"]).drop_duplicates().reset_index(drop=True)
raw.loc[raw["Age"] > 100, "Age"] = np.nan   # fix the outlier as missing

X_raw = raw.drop(columns=["Promoted", "EmpID"])
y_raw = raw["Promoted"]

num_features = ["Age", "Experience", "Salary", "Rating"]
cat_features = ["Department", "City"]

numeric_steps = Pipeline([
    ("fill_missing", SimpleImputer(strategy="median")),
    ("scale",        StandardScaler()),
])

categorical_steps = Pipeline([
    ("fill_missing", SimpleImputer(strategy="most_frequent")),
    ("encode",       OneHotEncoder(drop="first", handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("numbers", numeric_steps,     num_features),
    ("text",    categorical_steps, cat_features),
])
print("Preprocessor built.")
```

**Q23.** Fit pipeline on train split only, transform both, print transformed shapes.

```python
Xtr, Xte, ytr, yte = train_test_split(
    X_raw, y_raw, test_size=0.20, random_state=42
)

Xtr_ready = preprocessor.fit_transform(Xtr)   # learn + apply
Xte_ready = preprocessor.transform(Xte)       # apply only

print("Transformed train:", Xtr_ready.shape)
print("Transformed test: ", Xte_ready.shape)
```

**Q24.** Explain why we split before scaling.

**Answer:** We split the data *first* so the scaler never sees the test rows. `StandardScaler` learns the mean and standard deviation from the data it is fit on. If we scaled before splitting, those statistics would be computed using the test rows too, so information from the test set would leak into the training process. That makes the model's test score look better than it really is. The correct approach: split first, `fit` (or `fit_transform`) the scaler on the **training set only**, then `transform` the test set with those same learned values.
